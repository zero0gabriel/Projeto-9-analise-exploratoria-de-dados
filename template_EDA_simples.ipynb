Para cada coluna:


````python
# A coluna Idade_Media estava com uma vírgula ao invés de ponto, e por conta disso o tipo era String ao invés de float
df['Idade_Media'] = df['Idade_Media'].apply(lambda x: float(x.replace(',', '.')))
````

# 6. Distribuição das variáveis numéricas

Para cada variável numérica:

- Qual é o mínimo?

````python
df.info()
````

````python
colunas_numericas = []
for i in df.columns:
    if df[i].dtype != str:
        colunas_numericas.append(i)

colunas_numericas
````

````python
import numpy as np
colunas_numericas = []
for i in df.columns:
    if df[i].dtype != 'str':

        colunas_numericas.append(i)

df[colunas_numericas].agg({min, max, np.mean})
````
